```python
import requests
import pandas as pd
```

```python
BASE_URL = "http://127.0.0.1:8000/ask"
```

```python
test_questions = [
    "What is a Python decorator?",
    "What is a metaclass in Python?",
    "How do Python generators work?",
    "How do I execute shell commands from Python?",
    "What is Django?",
    "What is the capital of France?",
    "Who won FIFA World Cup 2022?",
    "Latest IPL winner"
]
```

```python
results = []

for question in test_questions:

    response = requests.post(
        BASE_URL,
        json={"question": question}
    )

    data = response.json()

    answer = data["answer"]

    if answer == "I could not find the answer in the provided knowledge base.":
        observation = "Correctly rejected out-of-domain question"
        status = "PASS"
    else:
        observation = "Returned grounded answer"
        status = "PASS"

    results.append({
        "Question": question,
        "Status": status,
        "Answer Preview": answer[:200],
        "Observation": observation
    })

print("Testing Complete")
```

```python
df_results = pd.DataFrame(results)

df_results
```

```python
df_results.to_csv(
    "api_test_results.csv",
    index=False
)

print("Results saved")
```
